```python

from pathlib import Path
from collections import Counter
import pandas as pd
import numpy as np
import gc

PROJECT_ROOT = Path("/content/drive/MyDrive/H-IDS")
RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"

label_counts = Counter()
file_summaries = []
column_reference = None
column_mismatches = []

print("DATA QUALITY AUDIT")
print("=" * 70)

for file in sorted(RAW_DATA_DIR.rglob("*.csv")):
    print(f"\nInspecting: {file.name}")

    try:
        df = pd.read_csv(
            file,
            low_memory=False,
            encoding="utf-8-sig"
        )

        # Normalize whitespace in column names for comparison only.
        normalized_columns = [
            str(col).strip() for col in df.columns
        ]

        if column_reference is None:
            column_reference = normalized_columns
        elif normalized_columns != column_reference:
            column_mismatches.append(file.name)

        # Locate the label column without assuming its leading space.
        label_matches = [
            col for col in df.columns
            if str(col).strip().lower() == "label"
        ]

        if len(label_matches) != 1:
            print("ERROR: Expected exactly one Label column.")
            del df
            gc.collect()
            continue

        label_col = label_matches[0]

        # Keep the original label text for the audit.
        labels = df[label_col].astype("string").str.strip()
        counts = labels.value_counts(dropna=False)

        for label, count in counts.items():
            label_counts[str(label)] += int(count)

        # Numeric data-quality checks.
        feature_df = df.drop(columns=[label_col])
        numeric_df = feature_df.select_dtypes(include=[np.number])

        missing_cells = int(df.isna().sum().sum())
        infinite_cells = int(
            np.isinf(numeric_df.to_numpy()).sum()
        )

        summary = {
            "file": file.name,
            "rows": len(df),
            "columns": len(df.columns),
            "missing_cells": missing_cells,
            "infinite_numeric_cells": infinite_cells,
            "unique_labels": labels.nunique(dropna=True)
        }
        file_summaries.append(summary)

        print("Rows:", f"{len(df):,}")
        print("Columns:", len(df.columns))
        print("Unique labels:", labels.nunique(dropna=True))
        print("Missing cells:", f"{missing_cells:,}")
        print("Infinite numeric cells:", f"{infinite_cells:,}")
        print("Label distribution:")
        print(counts.to_string())

        del df, feature_df, numeric_df, labels, counts
        gc.collect()

    except Exception as error:
        print("ERROR:", repr(error))

print("\n" + "=" * 70)
print("COMBINED AUDIT SUMMARY")
print("=" * 70)

print("Files successfully audited:", len(file_summaries))
print("Column mismatches:", column_mismatches or "None")

if file_summaries:
    summary_df = pd.DataFrame(file_summaries)
    print("\nPer-file summary:")
    display(summary_df)

print("\nLabel counts across all successfully audited files:")
for label, count in sorted(label_counts.items()):
    print(f"{label}: {count:,}")

print("\nAudit completed. Original CSV files were not modified.")
```

```python

from pathlib import Path
from collections import Counter
import pandas as pd
import numpy as np
import gc

PROJECT_ROOT = Path("/content/drive/MyDrive/H-IDS")
RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"

all_labels = Counter()
missing_label_rows = 0
column_mismatches = []
reference_columns = None

print("LABEL AND FEATURE-TYPE AUDIT")
print("=" * 70)

for file in sorted(RAW_DATA_DIR.rglob("*.csv")):
    print(f"\nFILE: {file.name}")

    header = pd.read_csv(file, nrows=0, encoding="utf-8-sig")
    original_columns = list(header.columns)
    normalized_columns = [str(c).strip() for c in original_columns]

    if reference_columns is None:
        reference_columns = normalized_columns
    elif normalized_columns != reference_columns:
        column_mismatches.append(file.name)

    label_matches = [
        c for c in original_columns
        if str(c).strip().lower() == "label"
    ]

    if len(label_matches) != 1:
        print("ERROR: Label column missing or ambiguous.")
        continue

    label_col = label_matches[0]
    feature_columns = [
        c for c in original_columns if c != label_col
    ]

    # Inspect representative feature types.
    sample = pd.read_csv(
        file,
        nrows=1000,
        encoding="utf-8-sig",
        low_memory=False
    )

    print("Feature count:", len(feature_columns))
    print("Sample numeric features:",
          int(sample[feature_columns].apply(
              pd.to_numeric, errors="coerce"
          ).notna().all(axis=0).sum()))
    print("Sample columns containing nonnumeric values:")
    converted = sample[feature_columns].apply(
        pd.to_numeric, errors="coerce"
    )
    for col in feature_columns:
        original_nonempty = sample[col].notna()
        failed = original_nonempty & converted[col].isna()
        if failed.any():
            print(f"  {str(col).strip()}: {int(failed.sum())} sample values")
    del sample, converted
    gc.collect()

    # Count labels and missing labels in chunks.
    file_labels = Counter()
    file_missing = 0

    for chunk in pd.read_csv(
        file,
        usecols=[label_col],
        chunksize=100_000,
        encoding="utf-8-sig",
        low_memory=False
    ):
        labels = chunk[label_col].astype("string").str.strip()
        file_missing += int(labels.isna().sum())
        labels = labels.dropna()

        for value, count in labels.value_counts().items():
            file_labels[str(value)] += int(count)

    all_labels.update(file_labels)
    missing_label_rows += file_missing

    print("Exact label representations:")
    for label, count in sorted(file_labels.items()):
        print(f"  {label!r}: {count:,}")
    print("Missing label rows:", file_missing)

print("\n" + "=" * 70)
print("FINAL CHECKS")
print("Column mismatches:", column_mismatches or "None")
print("Total rows with missing labels:", missing_label_rows)
print("Distinct labels:", len(all_labels))
print("\nLabels using exact Python representations:")
for label, count in sorted(all_labels.items()):
    print(f"{label!r}: {count:,}")

print("\nAudit finished. No source files were modified.")
```

```python

from pathlib import Path
from collections import Counter
import json
import uuid
import gc

import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq

# --------------------------------------------------
# 1. PATHS AND SAFETY CHECKS
# --------------------------------------------------
PROJECT_ROOT = Path("/content/drive/MyDrive/H-IDS")
RAW_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

OUTPUT_FILE = PROCESSED_DIR / "cic_ids2017_canonical.parquet"
MANIFEST_FILE = PROCESSED_DIR / "canonical_manifest.json"

CHUNK_SIZE = 25_000

PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

if OUTPUT_FILE.exists():
    raise FileExistsError(
        f"Output already exists: {OUTPUT_FILE}. "
        "Inspect it before creating another version."
    )

if MANIFEST_FILE.exists():
    raise FileExistsError(
        f"Manifest already exists: {MANIFEST_FILE}. "
        "Inspect it before creating another version."
    )

CSV_FILES = sorted(RAW_DIR.glob("*.csv"))

if len(CSV_FILES) != 8:
    raise ValueError(
        f"Expected 8 CSV files, found {len(CSV_FILES)}."
    )

TEMP_FILE = PROCESSED_DIR / (
    f"canonical_build_{uuid.uuid4().hex}.parquet"
)
TEMP_MANIFEST = PROCESSED_DIR / (
    f"canonical_manifest_{uuid.uuid4().hex}.json"
)

# --------------------------------------------------
# 2. DEFINE THE CANONICAL SCHEMA AND LABEL MAPPING
# --------------------------------------------------
first_header = pd.read_csv(
    CSV_FILES[0], nrows=0, encoding="utf-8-sig"
)

REFERENCE_COLUMNS = [
    str(col).strip() for col in first_header.columns
]

if len(REFERENCE_COLUMNS) != 79:
    raise ValueError(
        f"Expected 79 columns; got {len(REFERENCE_COLUMNS)}."
    )

if REFERENCE_COLUMNS.count("Label") != 1:
    raise ValueError("Expected exactly one Label column.")

FEATURE_COLUMNS = [
    col for col in REFERENCE_COLUMNS if col != "Label"
]

if len(FEATURE_COLUMNS) != 78:
    raise ValueError(
        f"Expected 78 features; got {len(FEATURE_COLUMNS)}."
    )

# Explicit mapping from observed source labels to clean names.
# Raw labels will also be retained in label_raw.
LABEL_MAP = {
    "BENIGN": "BENIGN",
    "Bot": "Bot",
    "DDoS": "DDoS",
    "DoS GoldenEye": "DoS GoldenEye",
    "DoS Hulk": "DoS Hulk",
    "DoS Slowhttptest": "DoS Slowhttptest",
    "DoS slowloris": "DoS slowloris",
    "FTP-Patator": "FTP-Patator",
    "Heartbleed": "Heartbleed",
    "Infiltration": "Infiltration",
    "PortScan": "PortScan",
    "SSH-Patator": "SSH-Patator",
    "Web Attack � Brute Force": "Web Attack Brute Force",
    "Web Attack � Sql Injection": "Web Attack SQL Injection",
    "Web Attack � XSS": "Web Attack XSS",
}

# --------------------------------------------------
# 3. COUNTERS AND WRITER
# --------------------------------------------------
writer = None
total_rows = 0
label_counts = Counter()
raw_label_counts = Counter()
source_counts = {}
missing_source_counts = Counter()
infinite_counts = Counter()
file_summaries = []
observed_labels = set()

try:
    for file in CSV_FILES:
        print(f"\nProcessing: {file.name}")

        header = pd.read_csv(
            file, nrows=0, encoding="utf-8-sig"
        )
        normalized_header = [
            str(col).strip() for col in header.columns
        ]

        if normalized_header != REFERENCE_COLUMNS:
            raise ValueError(
                f"Column mismatch detected in {file.name}."
            )

        original_label_col = next(
            col for col in header.columns
            if str(col).strip() == "Label"
        )

        file_rows = 0
        file_labels = Counter()
        source_row_start = 0

        for chunk in pd.read_csv(
            file,
            chunksize=CHUNK_SIZE,
            encoding="utf-8-sig",
            low_memory=False
        ):
            # Read labels before normalizing column names.
            raw_labels = (
                chunk[original_label_col]
                .astype("string")
                .str.strip()
            )

            if raw_labels.isna().any():
                raise ValueError(
                    f"Missing labels found in {file.name}."
                )

            unknown = (
                set(raw_labels.unique()) - set(LABEL_MAP.keys())
            )
            if unknown:
                raise ValueError(
                    f"Unmapped labels in {file.name}: {unknown}"
                )

            clean_labels = raw_labels.map(LABEL_MAP)

            if clean_labels.isna().any():
                raise ValueError(
                    f"Label mapping failed in {file.name}."
                )

            # Standardize feature column names.
            chunk.columns = [
                str(col).strip() for col in chunk.columns
            ]

            output = pd.DataFrame(index=range(len(chunk)))

            # Convert features to float64. Preserve rows.
            for col in FEATURE_COLUMNS:
                numeric = pd.to_numeric(
                    chunk[col], errors="coerce"
                )

                failed = chunk[col].notna() & numeric.isna()
                if failed.any():
                    examples = (
                        chunk.loc[failed, col]
                        .astype(str).head(5).tolist()
                    )
                    raise ValueError(
                        f"Nonnumeric values in {col!r}, "
                        f"{file.name}. Examples: {examples}"
                    )

                values = numeric.to_numpy(
                    dtype=np.float64, na_value=np.nan
                )

                missing_source_counts[col] += int(
                    np.isnan(values).sum()
                )

                infinite_mask = np.isinf(values)
                infinite_counts[col] += int(
                    infinite_mask.sum()
                )

                # Treat +inf and -inf as missing, without dropping rows.
                values[infinite_mask] = np.nan
                output[col] = values

            output["label_raw"] = raw_labels.astype(str).to_numpy()
            output["label"] = clean_labels.astype(str).to_numpy()
            output["source_file"] = file.name
            output["source_row"] = np.arange(
                source_row_start,
                source_row_start + len(chunk),
                dtype=np.int64
            )

            output = output[
                FEATURE_COLUMNS
                + ["label_raw", "label", "source_file", "source_row"]
            ]

            # Use the same schema for every written chunk.
            table = pa.Table.from_pandas(
                output, preserve_index=False
            )

            if writer is None:
                writer = pq.ParquetWriter(
                    TEMP_FILE,
                    table.schema,
                    compression="snappy"
                )
            elif not table.schema.equals(writer.schema):
                raise ValueError(
                    f"Output schema changed while processing {file.name}."
                )

            writer.write_table(
                table,
                row_group_size=CHUNK_SIZE
            )

            label_counts.update(
                output["label"].value_counts().to_dict()
            )
            raw_label_counts.update(
                output["label_raw"].value_counts().to_dict()
            )
            file_labels.update(
                output["label"].value_counts().to_dict()
            )
            observed_labels.update(output["label_raw"].unique())

            rows_in_chunk = len(output)
            total_rows += rows_in_chunk
            file_rows += rows_in_chunk
            source_row_start += rows_in_chunk

            del chunk, output, table, numeric, values
            gc.collect()

        source_counts[file.name] = file_rows
        file_summaries.append({
            "file": file.name,
            "rows": file_rows,
            "label_counts": dict(file_labels)
        })

        print("Rows processed:", f"{file_rows:,}")

    if writer is None:
        raise RuntimeError("No output data was written.")

    writer.close()
    writer = None

    # --------------------------------------------------
    # 4. VALIDATE THE COMPLETED TEMPORARY FILE
    # --------------------------------------------------
    expected_rows = sum(source_counts.values())
    parquet_file = pq.ParquetFile(TEMP_FILE)

    if parquet_file.metadata.num_rows != expected_rows:
        raise ValueError(
            f"Row-count mismatch: expected {expected_rows:,}, "
            f"found {parquet_file.metadata.num_rows:,}."
        )

    expected_columns = (
        FEATURE_COLUMNS
        + ["label_raw", "label", "source_file", "source_row"]
    )

    if parquet_file.schema.names != expected_columns:
        raise ValueError("Output column validation failed.")

    # Count output labels directly from the written file.
    verified_label_counts = Counter()
    verified_source_counts = Counter()
    verified_raw_counts = Counter()

    for batch in parquet_file.iter_batches(
        columns=["label", "label_raw", "source_file"],
        batch_size=100_000
    ):
        data = batch.to_pydict()
        verified_label_counts.update(
            Counter(data["label"])
        )
        verified_raw_counts.update(
            Counter(data["label_raw"])
        )
        verified_source_counts.update(
            Counter(data["source_file"])
        )

    if verified_label_counts != label_counts:
        raise ValueError("Output label counts do not match processing counts.")

    if verified_source_counts != Counter(source_counts):
        raise ValueError("Output source-file row counts do not match.")

    # --------------------------------------------------
    # 5. WRITE MANIFEST AND PUBLISH FINAL FILE
    # --------------------------------------------------
    manifest = {
        "dataset": "CIC-IDS2017 canonical processed dataset",
        "source_files": [file.name for file in CSV_FILES],
        "total_rows": total_rows,
        "feature_count": len(FEATURE_COLUMNS),
        "feature_columns": FEATURE_COLUMNS,
        "output_columns": expected_columns,
        "label_mapping": LABEL_MAP,
        "raw_label_counts": dict(raw_label_counts),
        "normalized_label_counts": dict(label_counts),
        "rows_by_source_file": source_counts,
        "missing_feature_values_before_inf_replacement":
            dict(missing_source_counts),
        "infinite_values_replaced_with_nan": dict(infinite_counts),
        "missing_label_rows": 0,
        "duplicate_rows_removed": 0,
        "imputation_performed": False,
        "split_created": False
    }

    with open(TEMP_MANIFEST, "w", encoding="utf-8") as f:
        json.dump(manifest, f, indent=2, ensure_ascii=False)

    # Publish only after successful output validation.
    TEMP_FILE.replace(OUTPUT_FILE)
    TEMP_MANIFEST.replace(MANIFEST_FILE)

    print("\nCANONICAL DATASET CREATED AND VALIDATED")
    print("=" * 60)
    print("Output:", OUTPUT_FILE)
    print("Manifest:", MANIFEST_FILE)
    print("Total rows:", f"{total_rows:,}")
    print("Feature columns:", len(FEATURE_COLUMNS))
    print("Total output columns:", len(expected_columns))
    print("Distinct labels:", len(label_counts))
    print("Duplicates removed: 0")
    print("Imputation performed: No")
    print("Source CSV files modified: No")

    print("\nNormalized label counts:")
    for label, count in sorted(label_counts.items()):
        print(f"{label}: {count:,}")

except Exception as exc:
    if writer is not None:
        writer.close()
        writer = None

    print("\nPROCESSING FAILED")
    print(repr(exc))
    print("Do not use any temporary file as the canonical dataset.")
    print("Original CSV files remain unchanged.")
    raise

finally:
    gc.collect()
```

```python

from pathlib import Path
import sqlite3
import numpy as np
import pandas as pd
import pyarrow.parquet as pq

PROJECT_ROOT = Path("/content/drive/MyDrive/H-IDS")
DATA_FILE = (
    PROJECT_ROOT / "data" / "processed"
    / "cic_ids2017_canonical.parquet"
)

DB_FILE = Path("/content/hids_duplicate_audit.sqlite")

# Start a fresh audit database.
if DB_FILE.exists():
    DB_FILE.unlink()

conn = sqlite3.connect(str(DB_FILE))
conn.execute("PRAGMA journal_mode = OFF")
conn.execute("PRAGMA synchronous = OFF")
conn.execute("PRAGMA temp_store = FILE")

conn.execute("""
    CREATE TABLE row_hashes (
        feature_hash INTEGER NOT NULL,
        label TEXT NOT NULL
    )
""")

parquet = pq.ParquetFile(DATA_FILE)
feature_columns = [
    col for col in parquet.schema.names
    if col not in {
        "label", "label_raw", "source_file", "source_row"
    }
]

print("Building disk-based duplicate audit index...")

try:
    for batch_number, batch in enumerate(
        parquet.iter_batches(
            columns=feature_columns + ["label"],
            batch_size=50_000
        ),
        start=1
    ):
        df = batch.to_pandas()

        # Hash only the 78 feature values, not the label or source metadata.
        hashes = (
            pd.util.hash_pandas_object(
                df[feature_columns], index=False
            )
            .to_numpy(dtype=np.uint64)
            .view(np.int64)
        )

        rows = zip(hashes.tolist(), df["label"].tolist())

        conn.executemany(
            "INSERT INTO row_hashes VALUES (?, ?)",
            rows
        )
        conn.commit()

        if batch_number % 10 == 0:
            print(f"Processed batches: {batch_number}")

    print("\nCalculating duplicate statistics...")

    conn.execute("""
        CREATE INDEX idx_feature_hash
        ON row_hashes(feature_hash)
    """)
    conn.commit()

    total_rows = conn.execute(
        "SELECT COUNT(*) FROM row_hashes"
    ).fetchone()[0]

    repeated = conn.execute("""
        SELECT COUNT(*), COALESCE(SUM(n), 0)
        FROM (
            SELECT COUNT(*) AS n
            FROM row_hashes
            GROUP BY feature_hash
            HAVING COUNT(*) > 1
        )
    """).fetchone()

    conflicting = conn.execute("""
        SELECT COUNT(*), COALESCE(SUM(n), 0)
        FROM (
            SELECT COUNT(*) AS n
            FROM row_hashes
            GROUP BY feature_hash
            HAVING COUNT(DISTINCT label) > 1
        )
    """).fetchone()

    print("\nDUPLICATE LABEL AUDIT")
    print("=" * 55)
    print("Rows indexed:", f"{total_rows:,}")
    print("Hashes occurring more than once:", f"{repeated[0]:,}")
    print("Rows under repeated hashes:", f"{repeated[1]:,}")
    print("Hashes associated with multiple labels:", f"{conflicting[0]:,}")
    print("Rows under potentially conflicting hashes:", f"{conflicting[1]:,}")

finally:
    conn.close()

print("\nAudit finished. Temporary database:", DB_FILE)
print("The canonical dataset has not been changed.")
```

```python

from pathlib import Path
from collections import Counter, defaultdict
import sqlite3
import numpy as np
import pandas as pd
import pyarrow.parquet as pq

PROJECT_ROOT = Path("/content/drive/MyDrive/H-IDS")
DATA_FILE = (
    PROJECT_ROOT / "data" / "processed"
    / "cic_ids2017_canonical.parquet"
)
DB_FILE = Path("/content/hids_duplicate_audit.sqlite")

if not DATA_FILE.exists():
    raise FileNotFoundError(DATA_FILE)

if not DB_FILE.exists():
    raise FileNotFoundError(
        "The temporary SQLite database is missing. "
        "Rerun the previous duplicate audit first."
    )

parquet = pq.ParquetFile(DATA_FILE)
feature_columns = [
    col for col in parquet.schema.names
    if col not in {
        "label", "label_raw", "source_file", "source_row"
    }
]

# Retrieve only hashes previously flagged as having multiple labels.
conn = sqlite3.connect(str(DB_FILE))
candidate_hashes = {
    row[0]
    for row in conn.execute("""
        SELECT feature_hash
        FROM row_hashes
        GROUP BY feature_hash
        HAVING COUNT(DISTINCT label) > 1
    """)
}
conn.close()

print("Candidate hashes:", f"{len(candidate_hashes):,}")
print("Scanning dataset to verify exact feature values...")

# Map each exact feature vector to its labels and source records.
exact_groups = defaultdict(lambda: {
    "labels": Counter(),
    "examples": []
})

rows_checked = 0

for batch in parquet.iter_batches(
    columns=feature_columns + [
        "label", "source_file", "source_row"
    ],
    batch_size=50_000
):
    df = batch.to_pandas()

    hashes = (
        pd.util.hash_pandas_object(
            df[feature_columns], index=False
        )
        .to_numpy(dtype=np.uint64)
        .view(np.int64)
    )

    mask = np.isin(
        hashes,
        np.fromiter(candidate_hashes, dtype=np.int64)
    )

    if not mask.any():
        continue

    candidates = df.loc[mask]

    for row in candidates.itertuples(index=False, name=None):
        values = row[:len(feature_columns)]
        label = row[len(feature_columns)]
        source_file = row[len(feature_columns) + 1]
        source_row = row[len(feature_columns) + 2]

        # Canonicalize NaN values for exact row comparison.
        signature = tuple(
            None if pd.isna(value) else float(value)
            for value in values
        )

        group = exact_groups[signature]
        group["labels"][label] += 1

        if len(group["examples"]) < 5:
            group["examples"].append({
                "label": label,
                "source_file": source_file,
                "source_row": int(source_row)
            })

    rows_checked += len(candidates)

    if rows_checked and rows_checked % 10_000 < len(candidates):
        print("Candidate records checked:", f"{rows_checked:,}")

verified_conflicts = [
    (signature, group)
    for signature, group in exact_groups.items()
    if len(group["labels"]) > 1
]

print("\nEXACT FEATURE-VECTOR VERIFICATION")
print("=" * 60)
print("Candidate records examined:", f"{rows_checked:,}")
print("Distinct exact feature vectors checked:", f"{len(exact_groups):,}")
print(
    "Exact feature vectors with conflicting labels:",
    f"{len(verified_conflicts):,}"
)

conflicting_records = sum(
    sum(group["labels"].values())
    for _, group in verified_conflicts
)

print(
    "Records belonging to verified conflicting vectors:",
    f"{conflicting_records:,}"
)

print("\nFirst 10 verified conflicting groups:")
for index, (_, group) in enumerate(verified_conflicts[:10], start=1):
    print(f"\nGroup {index}")
    print("Label counts:", dict(group["labels"]))
    print("Example source records:", group["examples"])

print("\nVerification completed. Canonical dataset unchanged.")
```

```python

from pathlib import Path
import json

import numpy as np
import pandas as pd
import pyarrow.parquet as pq
from sklearn.model_selection import GroupShuffleSplit

PROJECT_ROOT = Path("/content/drive/MyDrive/H-IDS")
DATA_FILE = (
    PROJECT_ROOT / "data" / "processed"
    / "cic_ids2017_canonical.parquet"
)
SPLIT_DIR = PROJECT_ROOT / "data" / "splits"
SPLIT_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_FILE = SPLIT_DIR / "duplicate_aware_split.parquet"
SUMMARY_FILE = SPLIT_DIR / "duplicate_aware_split_summary.json"

if OUTPUT_FILE.exists() or SUMMARY_FILE.exists():
    raise FileExistsError(
        "Split output already exists. Inspect it before overwriting."
    )

parquet = pq.ParquetFile(DATA_FILE)
feature_columns = [
    c for c in parquet.schema.names
    if c not in {
        "label", "label_raw", "source_file", "source_row"
    }
]

row_hash_parts = []
label_parts = []
row_id_parts = []

print("Computing stable row hashes and collecting labels...")

row_id_start = 0

for batch in parquet.iter_batches(
    columns=feature_columns + ["label"],
    batch_size=50_000
):
    df = batch.to_pandas()

    hashes = (
        pd.util.hash_pandas_object(
            df[feature_columns], index=False
        )
        .to_numpy(dtype=np.uint64)
    )

    n = len(df)
    row_hash_parts.append(hashes)
    label_parts.append(df["label"].astype(str).to_numpy())
    row_id_parts.append(
        np.arange(row_id_start, row_id_start + n, dtype=np.int64)
    )
    row_id_start += n

row_hashes = np.concatenate(row_hash_parts)
labels = np.concatenate(label_parts)
row_ids = np.concatenate(row_id_parts)

assert len(row_ids) == parquet.metadata.num_rows
assert len(row_hashes) == len(labels) == len(row_ids)

print("Rows collected:", f"{len(row_ids):,}")
print("Distinct feature hashes:", f"{len(np.unique(row_hashes)):,}")

# First hold out approximately 20% of groups for testing.
outer_split = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)
trainval_idx, test_idx = next(
    outer_split.split(row_ids, labels, groups=row_hashes)
)

# From the remaining 80%, hold out 12.5% for validation.
# This targets approximately 10% of the complete dataset.
inner_split = GroupShuffleSplit(
    n_splits=1,
    test_size=0.125,
    random_state=43
)
train_rel, val_rel = next(
    inner_split.split(
        row_ids[trainval_idx],
        labels[trainval_idx],
        groups=row_hashes[trainval_idx]
    )
)

train_idx = trainval_idx[train_rel]
val_idx = trainval_idx[val_rel]

split = np.full(len(row_ids), "train", dtype=object)
split[val_idx] = "validation"
split[test_idx] = "test"

# Verify that no feature hash occurs in more than one split.
hash_splits = pd.DataFrame({
    "feature_hash": row_hashes,
    "split": split
}).groupby("feature_hash")["split"].nunique()

assert hash_splits.max() == 1, (
    "Duplicate feature hashes crossed split boundaries."
)

split_df = pd.DataFrame({
    "row_id": row_ids,
    "feature_hash": row_hashes,
    "label": labels,
    "split": split
})

split_df.to_parquet(OUTPUT_FILE, index=False)

summary = {
    "seed_test": 42,
    "seed_validation": 43,
    "method": "GroupShuffleSplit using feature-row hashes",
    "group_leakage_check_passed": True,
    "total_rows": int(len(row_ids)),
    "split_counts": {
        str(k): int(v)
        for k, v in pd.Series(split).value_counts().items()
    },
    "class_counts_by_split": {
        part: {
            str(k): int(v)
            for k, v in pd.Series(labels[split == part])
                .value_counts().items()
        }
        for part in ["train", "validation", "test"]
    },
    "classes_missing_by_split": {
        part: sorted(
            set(np.unique(labels)) -
            set(labels[split == part])
        )
        for part in ["train", "validation", "test"]
    }
}

with open(SUMMARY_FILE, "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2, ensure_ascii=False)

print("\nSPLIT SIZES")
print("=" * 45)
for part in ["train", "validation", "test"]:
    count = int((split == part).sum())
    print(f"{part}: {count:,} ({count / len(split):.2%})")

print("\nCLASSES MISSING FROM EACH SPLIT")
for part, missing in summary["classes_missing_by_split"].items():
    print(f"{part}: {missing if missing else 'None'}")

print("\nCLASS COUNTS BY SPLIT")
for part in ["train", "validation", "test"]:
    print(f"\n{part.upper()}")
    print(
        pd.Series(labels[split == part])
        .value_counts()
        .sort_index()
        .to_string()
    )

print("\nDuplicate-aware split saved:")
print(OUTPUT_FILE)
print(SUMMARY_FILE)
print("Cross-split repeated-hash check: PASSED")
```

```python

from pathlib import Path
from collections import defaultdict, Counter
import pandas as pd
import pyarrow.parquet as pq

PROJECT_ROOT = Path("/content/drive/MyDrive/H-IDS")
DATA_FILE = (
    PROJECT_ROOT / "data" / "processed"
    / "cic_ids2017_canonical.parquet"
)
SPLIT_FILE = (
    PROJECT_ROOT / "data" / "splits"
    / "duplicate_aware_split.parquet"
)

rare_classes = [
    "Heartbleed",
    "Infiltration",
    "Web Attack SQL Injection",
    "Web Attack XSS",
    "Web Attack Brute Force",
    "Bot",
]

# Read the split metadata and align it to the canonical row order.
split_df = pd.read_parquet(SPLIT_FILE).sort_values("row_id")
assert split_df["row_id"].is_unique
assert split_df["row_id"].iloc[0] == 0
assert split_df["row_id"].iloc[-1] == len(split_df) - 1

split_names = split_df["split"].to_numpy()
split_hashes = split_df["feature_hash"].to_numpy()

parquet = pq.ParquetFile(DATA_FILE)
counts = defaultdict(Counter)
group_labels = defaultdict(set)

row_id = 0

for batch in parquet.iter_batches(
    columns=["label"],
    batch_size=50_000
):
    labels = batch.column(0).to_pylist()
    batch_len = len(labels)

    for offset, label in enumerate(labels):
        idx = row_id + offset
        part = split_names[idx]
        feature_hash = int(split_hashes[idx])

        if label in rare_classes:
            counts[(label, part)]["records"] += 1
            counts[(label, part)]["feature_groups"] += 0

        group_labels[feature_hash].add(label)

    row_id += batch_len

assert row_id == len(split_df)

# Count distinct feature groups for each rare label and split.
seen = set()
row_id = 0

for batch in parquet.iter_batches(
    columns=["label"],
    batch_size=50_000
):
    labels = batch.column(0).to_pylist()

    for offset, label in enumerate(labels):
        idx = row_id + offset
        if label not in rare_classes:
            continue

        key = (
            label,
            split_names[idx],
            int(split_hashes[idx])
        )

        if key not in seen:
            counts[(label, split_names[idx])]["feature_groups"] += 1
            seen.add(key)

    row_id += len(labels)

print("RARE-CLASS GROUP AUDIT")
print("=" * 65)

for label in rare_classes:
    print(f"\n{label}")
    for part in ["train", "validation", "test"]:
        c = counts[(label, part)]
        print(
            f"  {part:10s} records={c['records']:5,d}, "
            f"feature_groups={c['feature_groups']:5,d}"
        )

print("\nConflicting-label groups among the rare classes:")
conflicting_hashes = {
    h for h, labels in group_labels.items()
    if len(labels) > 1
}
print("Groups with multiple labels:", f"{len(conflicting_hashes):,}")
print("Audit completed. No data has been modified.")
```

```python

from pathlib import Path
import gc

import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq

PROJECT_ROOT = Path("/content/drive/MyDrive/H-IDS")
DATA_FILE = (
    PROJECT_ROOT / "data" / "processed"
    / "cic_ids2017_canonical.parquet"
)
SPLIT_FILE = (
    PROJECT_ROOT / "data" / "splits"
    / "duplicate_aware_split.parquet"
)
SPLIT_DIR = PROJECT_ROOT / "data" / "splits"

OUTPUTS = {
    "train": SPLIT_DIR / "train.parquet",
    "validation": SPLIT_DIR / "validation.parquet",
    "test": SPLIT_DIR / "test.parquet",
}

for path in OUTPUTS.values():
    if path.exists():
        raise FileExistsError(
            f"Output already exists: {path}. Inspect it before rerunning."
        )

split_df = pd.read_parquet(SPLIT_FILE).sort_values("row_id")

assert split_df["row_id"].is_unique
assert len(split_df) == pq.ParquetFile(DATA_FILE).metadata.num_rows
assert np.array_equal(
    split_df["row_id"].to_numpy(),
    np.arange(len(split_df), dtype=np.int64)
)

split_assignments = split_df["split"].to_numpy()
source = pq.ParquetFile(DATA_FILE)

writers = {name: None for name in OUTPUTS}
counts = {name: 0 for name in OUTPUTS}

try:
    row_start = 0

    for batch in source.iter_batches(batch_size=50_000):
        df = batch.to_pandas()
        row_end = row_start + len(df)

        assignments = split_assignments[row_start:row_end]

        for name, output_path in OUTPUTS.items():
            mask = assignments == name
            if not mask.any():
                continue

            subset = df.loc[mask].reset_index(drop=True)
            table = pa.Table.from_pandas(
                subset, preserve_index=False
            )

            if writers[name] is None:
                writers[name] = pq.ParquetWriter(
                    output_path,
                    table.schema,
                    compression="snappy"
                )
            elif not table.schema.equals(writers[name].schema):
                raise ValueError(
                    f"Schema mismatch while writing {name}."
                )

            writers[name].write_table(table)
            counts[name] += len(subset)
            del subset, table

        row_start = row_end
        del df, batch
        gc.collect()

    assert row_start == len(split_assignments)

finally:
    for writer in writers.values():
        if writer is not None:
            writer.close()

print("SPLIT FILES CREATED")
print("=" * 50)

for name, path in OUTPUTS.items():
    saved_rows = pq.ParquetFile(path).metadata.num_rows
    assert saved_rows == counts[name]
    print(f"{name:10s}: {saved_rows:,} rows")
    print(f"           {path}")

assert sum(counts.values()) == len(split_assignments)
print("\nTotal records preserved:", f"{sum(counts.values()):,}")
print("Canonical dataset modified: No")
```

```python

from pathlib import Path
import pandas as pd
import pyarrow.parquet as pq

PROJECT_ROOT = Path("/content/drive/MyDrive/H-IDS")
SPLIT_DIR = PROJECT_ROOT / "data" / "splits"

expected_rows = {
    "train": 1_990_486,
    "validation": 285_201,
    "test": 555_056,
}

for split_name, expected in expected_rows.items():
    path = SPLIT_DIR / f"{split_name}.parquet"
    assert path.exists(), f"Missing split file: {path}"

    parquet = pq.ParquetFile(path)
    assert parquet.metadata.num_rows == expected

    print(f"\n{split_name.upper()}")
    print("Rows:", f"{parquet.metadata.num_rows:,}")
    print("Columns:", parquet.schema.names[-4:])

    label_counts = {}

    for batch in parquet.iter_batches(
        columns=["label"],
        batch_size=100_000
    ):
        counts = pd.Series(batch.column(0).to_pylist()).value_counts()
        for label, count in counts.items():
            label_counts[label] = label_counts.get(label, 0) + int(count)

    benign = label_counts.get("BENIGN", 0)
    attacks = sum(label_counts.values()) - benign

    print("BENIGN (target 0):", f"{benign:,}")
    print("ATTACK (target 1):", f"{attacks:,}")
    print("Total:", f"{sum(label_counts.values()):,}")

    assert benign + attacks == expected

print("\nAll three split files passed the row-count checks.")
```
